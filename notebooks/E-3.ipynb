E-3: Il **Gradient checking** è un modo per verificare che il gradiente scritto a mano sia corretto, confrontandolo con un'approssimazione numerica
che non richiede di aver capito la formula, solo di saper valutare la funzione di errore: (L(θ+ε) − L(θ−ε))/(2ε), con ε = 10⁻⁵. Il suo gradiente
analitico e questa approssimazione devono coincidere a 6–7 cifre decimali. Se non coincidono, il bug è nel gradiente scritto a mano, non
nell'approssimazione. Questa tecnica Le farà risparmiare giorni interi più avanti: la impari adesso, su un caso in cui può verificare tutto a occhio.

```python
import numpy as np

# Generazione dei dati
np.random

N = 200

x = np.random.uniform(-10, 10, N)

rumore = np.random.normal(0, 5, N)

y = 3 * x + 1 + rumore

#parametri da verificare
m = 5
q = 3

epsilon = 1e-5

#calcolo del gradiente analitico
N = len(x)

y_pred = m * x + q

errore = y_pred - y

#Derivata rispetto a m
grad_m_analitico = (1/(N))*np.sum(errore*x)
#Derivata rispetto a q
grad_q_analitico = (1/(N))*np.sum(errore)

#gradiente analitico
gradiente_analitico = np.array([grad_m_analitico, grad_q_analitico])

#calcolo del gradiente numerico

#Derivata rispetto a m
m_con_epsilon = (m + epsilon) * x + q
costo_m_piu_epsilon = (1/(2*N))*np.sum((m_con_epsilon-y)**2)
m_meno_epsilon = (m - epsilon) * x + q
costo_m_meno_epsilon = (1/(2*N))*np.sum((m_meno_epsilon-y)**2)

grad_m_numerico = (costo_m_piu_epsilon - costo_m_meno_epsilon)/(2*epsilon)

#Derivata rispetto a q
q_con_epsilon = m * x + (q+ epsilon)
costo_q_piu_epsilon = (1/(2*N))*np.sum((q_con_epsilon-y)**2)
q_meno_epsilon = m * x + (q - epsilon)
costo_q_meno_epsilon = (1/(2*N))*np.sum((q_meno_epsilon-y)**2)

grad_q_numerico = (costo_q_piu_epsilon - costo_q_meno_epsilon)/(2*epsilon)

#gradiente numerico
gradiente_numerico = np.array([grad_m_numerico, grad_q_numerico])


print(f"Gradiente analitico: {gradiente_analitico}")
print(f"Gradiente numerico: {gradiente_numerico}")

#trovo la differenza tra i due gradienti per verificare l'errore
differenza = np.abs(gradiente_analitico - gradiente_numerico)

print(f"Differenza tra i gradienti: {differenza}")

#controllo la grandezza dell'errore per verificare che il calcolo sia attendibile
if(np.all(differenza<1e-6)):
  print("Gradiente corretto")
else:
  print("Gradiente errato")
```
